````python
import json
from pathlib import Path

import pandas as pd

root_dir = Path.cwd()
if not (root_dir / "artifacts").exists():
    root_dir = root_dir.parent

pytorch_dir = root_dir / "artifacts/pytorch"

with open(pytorch_dir / "metrics.json", encoding="utf-8") as file:
    pytorch_metrics = json.load(file)

pytorch_metrics["validation"], pytorch_metrics["test"]
````

````python
history = pd.read_json(pytorch_dir / "training_history.json")
history.tail()
````

````python
history[["train_loss", "validation_loss"]].plot(title="PyTorch MLP Training History")
````
