# 01 Data Exploration

Explore the raw NFL schedules and weekly team statistics loaded by `src/data/load_raw_data.py`.

Reproduce the raw data files from the project root:

```bash
.venv/bin/python src/data/load_raw_data.py
```

````python
from pathlib import Path

import pandas as pd

root_dir = Path.cwd()
if not (root_dir / "data").exists():
    root_dir = root_dir.parent

raw_dir = root_dir / "data/raw"
games = pd.read_parquet(raw_dir / "games.parquet")
team_stats = pd.read_parquet(raw_dir / "team_stats.parquet")

games.shape, team_stats.shape
````

````python
games[["season", "week", "game_type", "gameday", "home_team", "away_team", "home_score", "away_score"]].head()
````

````python
games["season"].agg(["min", "max"]), games["game_type"].value_counts()
````
