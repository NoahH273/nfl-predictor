# 02 Baseline Models

Review the trained logistic regression and XGBoost artifacts. Model training remains reproducible from the source scripts.

Reproduce the baseline model artifacts from the project root:

```bash
.venv/bin/python src/models/train_logistic_regression.py
.venv/bin/python src/models/train_xgboost.py
```

````python
import json
from pathlib import Path

import pandas as pd

root_dir = Path.cwd()
if not (root_dir / "artifacts").exists():
    root_dir = root_dir.parent

artifacts = root_dir / "artifacts"

with open(artifacts / "logistic_regression" / "metrics.json", encoding="utf-8") as file:
    logistic_metrics = json.load(file)

with open(artifacts / "xgboost" / "metrics.json", encoding="utf-8") as file:
    xgboost_metrics = json.load(file)

logistic_metrics["test"], xgboost_metrics["test"]
````

````python
comparison = pd.DataFrame(
    [
        {"model": "Logistic Regression", **logistic_metrics["test"]},
        {"model": "XGBoost", **xgboost_metrics["test"]},
    ]
)
comparison
````
